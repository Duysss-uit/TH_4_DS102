```python
import numpy as np
from pathlib import Path
from PIL import Image

def load_dataset(directory, image_size=(512, 512)):
    X = []
    y = []

    labels = {
        "NORMAL": -1,
        "PNEUMONIA": 1
    }

    for class_name, label in labels.items():

        folder = Path(directory) / class_name

        for file in folder.glob("*"):

            if file.suffix.lower() not in (
                ".jpeg", ".jpg", ".png"
            ):
                continue

            with Image.open(file) as img:
                img = img.convert("L")
                img = img.resize(image_size)

                img = np.asarray(
                    img, dtype=np.float32
                ) / 255.0

                X.append(img.flatten())
                y.append(label)

    return np.array(X), np.array(y)
```

```python
dataset_path = Path("/kaggle/input/datasets/paultimothymooney/chest-xray-pneumonia/chest_xray")

X_train, y_train = load_dataset(dataset_path / "train")
X_test, y_test = load_dataset(dataset_path / "test")
X_val, y_val = load_dataset(dataset_path / "val")

print(X_train.shape)
print(y_train.shape)

print(X_test.shape)
print(y_test.shape)
```

Assignment 1

```python
from SVM import SVM

hard_svm = SVM()

hard_svm.fit(
    X=X_train, y=y_train,
    optimizer="smo",
    margin="hard"
)
```

Assignment 2

```python
soft_svm = SVM(C=1.0)

soft_svm.fit(
    X_train, y_train,
    optimizer="smo",
    margin="soft"
)
```

Assignment 3

```python
sgd_svm = SVM(C=1.0)

sgd_svm.fit(
    X_train, y_train,
    optimizer="sgd",
    margin="soft",
    learning_rate=1e-7,
    epochs=10
)
```

```python
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score
)

def evaluate(model, X_test, y_test):

    y_pred = model.predict(X_test)

    print("Precision:", precision_score(y_test, y_pred,
                                       zero_division=0))
    print("Recall:", recall_score(y_test, y_pred,
                                 zero_division=0))
    print("F1-score:", f1_score(y_test, y_pred,
                              zero_division=0))


evaluate(hard_svm, X_test, y_test)
evaluate(soft_svm, X_test, y_test)
evaluate(sgd_svm, X_test, y_test)
```
